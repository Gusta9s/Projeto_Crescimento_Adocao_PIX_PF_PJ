Imports nescessarios

```python
import csv
```

Exploracao de dados

```python
array_inicial = []

with open('/home/gustavo/Projeto_base_pix_percentual_acomodacao/Projeto_Crescimento_Adocao_PIX_PF_PJ_FULLPLAYERS/data_raw/arq.csv', 'r') as file:
    reader = csv.DictReader(file, delimiter=',')
    for row in reader:
        array_inicial.append(row)

print(array_inicial[0])
```

Divisao Funcional entre PF e PJ

```python
array_inicial_pf = []
array_inicial_pj = []

with open('/home/gustavo/Projeto_base_pix_percentual_acomodacao/Projeto_Crescimento_Adocao_PIX_PF_PJ_FULLPLAYERS/resources/name_column_pf', 'r') as file:
    name_column_pf = file.read()

with open('/home/gustavo/Projeto_base_pix_percentual_acomodacao/Projeto_Crescimento_Adocao_PIX_PF_PJ_FULLPLAYERS/resources/name_column_pj', 'r') as file:
    name_column_pj = file.read()

with open('/home/gustavo/Projeto_base_pix_percentual_acomodacao/Projeto_Crescimento_Adocao_PIX_PF_PJ_FULLPLAYERS/resources/name_column_pf_final', 'r') as file:
    name_column_pf_final = file.read()

with open('/home/gustavo/Projeto_base_pix_percentual_acomodacao/Projeto_Crescimento_Adocao_PIX_PF_PJ_FULLPLAYERS/resources/name_column_pj_final', 'r') as file:
    name_column_pj_final = file.read()

for row in array_inicial:
    row[name_column_pf_final] = ''
    row[name_column_pj_final] = ''


for row in array_inicial:
    array_inicial_pf.append(row.get(name_column_pf, 'DESCONHECIDO'))
    array_inicial_pj.append(row.get(name_column_pj, 'DESCONHECIDO'))

print(array_inicial_pf[0:5])
print(f'\n {array_inicial_pj[0:5]}')
```

Aplicando calculo de crescimento em percentual por mes

```python
array_final_pf = []
array_final_pj = []

for i in range(0, len(array_inicial_pf) - 1):
    num1 = int(array_inicial_pf[i])
    num2 = int(array_inicial_pf[i + 1])
    percent_pf = ((num2 - num1) / num1) * 100
    array_final_pf.append(percent_pf)

for i in range(0, len(array_inicial_pj) - 1):
    num1 = int(array_inicial_pj[i])
    num2 = int(array_inicial_pj[i + 1])
    percent_pj = ((num2 - num1) / num1) * 100
    array_final_pj.append(percent_pj)

array_final_pf.insert(0, 0.0)
array_final_pj.insert(0, 0.0)

array_final_pf.pop(-1)
array_final_pj.pop(-1)

print(f'Percentual de crescimento de contas PF: {array_final_pf[0:5]}')
print(f'Percentual de crescimento de contas PJ: {array_final_pj[0:5]}')

print(f'LenghtA {len(array_final_pf)}, LenghtB {len(array_final_pj)}')
```

Geracao de arquivo .csv final

```python
array_final = []

array_final.append(list(array_inicial[0].keys()))

with open('/home/gustavo/Projeto_base_pix_percentual_acomodacao/Projeto_Crescimento_Adocao_PIX_PF_PJ_FULLPLAYERS/data_processed/arq.csv', 'w') as file:
    writer = csv.writer(file, delimiter=',')
    writer.writerow(array_final[0])

    for i in range(0, len(array_inicial) - 1):
        array_inicial[i][name_column_pf_final] = array_final_pf[i]
        array_inicial[i][name_column_pj_final] = array_final_pj[i]
        array_final.append(list(array_inicial[i].values()))
        writer.writerow(array_final[i+1])
```
